# NEXTBUY: Data Preprocessing

## Part 1: Loading & Merging Datasets

This notebook loads and preprocesses the 5 CSV datasets from the `datasets/` folder.

```python
import pandas as pd
import numpy as np
from pathlib import Path
```

## Configuration

```python
DATA_PATH = Path("datasets/")
print(f"Data path: {DATA_PATH.absolute()}")
```

## Load Datasets

```python
print("Loading datasets...")

aisles = pd.read_csv(DATA_PATH / "aisles.csv")
departments = pd.read_csv(DATA_PATH / "departments.csv")
products = pd.read_csv(DATA_PATH / "products.csv")
orders = pd.read_csv(DATA_PATH / "orders.csv")
order_products = pd.read_csv(DATA_PATH / "order_products.csv")

print("Datasets loaded successfully!")
```

## Dataset Overview

```python
datasets = {
    "Aisles": aisles,
    "Departments": departments,
    "Products": products,
    "Orders": orders,
    "Order Products": order_products
}

for name, df in datasets.items():
    print(f"\n{name}: {df.shape[0]:,} rows, {df.shape[1]} columns")
```

## Preview Data

```python
print("Orders sample:")
display(orders.head(3))

print("Order Products sample:")
display(order_products.head(3))
```

## Merge Datasets

```python
# Step 1: Merge products with aisles and departments
products_merged = products.merge(aisles, on="aisle_id", how="left")
products_merged = products_merged.merge(departments, on="department_id", how="left")
print(f"Products merged: {products_merged.shape}")
```

```python
# Step 2: Merge order_products with products
order_products_merged = order_products.merge(products_merged, on="product_id", how="left")
print(f"Order products merged: {order_products_merged.shape}")
```

```python
# Step 3: Merge with orders
full_data = order_products_merged.merge(orders, on="order_id", how="left")
print(f"Full dataset: {full_data.shape}")
```

## Handle Missing Values

```python
print("Missing values:")
missing = full_data.isnull().sum()
print(missing[missing > 0])
```

```python
# Fill days_since_prior_order: NaN means first order
full_data["days_since_prior_order"] = full_data["days_since_prior_order"].fillna(-1)
print(f"Missing values after handling: {full_data.isnull().sum().sum()}")
```

## Data Validation

```python
print("Data Validation:")
print(f"- Unique orders: {full_data['order_id'].nunique():,}")
print(f"- Unique users: {full_data['user_id'].nunique():,}")
print(f"- Unique products: {full_data['product_id'].nunique():,}")
print(f"- Reorder rate: {full_data['reordered'].mean()*100:.2f}%")
```

## Summary

```python
print("\n" + "="*50)
print("DATA PREPROCESSING COMPLETE")
print("="*50)
print(f"Final shape: {full_data.shape}")
print(f"Columns: {list(full_data.columns)}")
```
